```python
# ==========================================================
# Q-LEARNING FROM SCRATCH EN PYTHON
# ==========================================================
#
# Q-LEARNING :
# ----------------------------------------------------------
# Algorithme de Reinforcement Learning
# (apprentissage par renforcement)
#
# ----------------------------------------------------------
# IDÉE PRINCIPALE :
# ----------------------------------------------------------
#
# Un agent apprend en interagissant avec un environnement.
#
# Il ne reçoit pas directement la bonne réponse,
# mais apprend par :
#
# -> récompenses
# -> pénalités
#
# ----------------------------------------------------------
# OBJECTIF :
# ----------------------------------------------------------
#
# apprendre la meilleure action à faire
# dans chaque état.
#
# ----------------------------------------------------------
# CONCEPT IMPORTANT :
# ----------------------------------------------------------
#
# Q(s, a) :
#
# valeur de faire action a dans état s
#
# ----------------------------------------------------------
# EXEMPLES :
# ----------------------------------------------------------
#
# robot dans un labyrinthe
# jeu vidéo
# navigation GPS
#
# ==========================================================

# ==========================================================
# IMPORTATION DES LIBRARIES
# ==========================================================

# numpy :
# calculs mathématiques
import numpy as np

# random :
# choix aléatoire (exploration)
import random

# ==========================================================
# 1. ENVIRONNEMENT (GRID WORLD)
# ==========================================================
#
# On crée un petit monde en grille :
#
# taille : 4x4
#
# ----------------------------------------------------------
# OBJECTIF :
# ----------------------------------------------------------
#
# aller de START (0,0)
# vers GOAL (3,3)
#
# ==========================================================

grid_size = 4

# état objectif (case finale)
goal_state = (3, 3)

# ----------------------------------------------------------
# actions possibles
# ----------------------------------------------------------
#
# up    -> monter
# down  -> descendre
# left  -> gauche
# right -> droite
#
# ==========================================================

actions = ["up", "down", "left", "right"]

# ==========================================================
# 2. INITIALISATION Q-TABLE
# ==========================================================
#
# Q-table = mémoire de l'agent
#
# elle stocke :
#
# Q(state, action)
#
# ==========================================================
#
# structure :
#
# Q[(x,y)][action] = valeur
#
# ==========================================================

Q = {}

# ----------------------------------------------------------
# initialisation pour chaque état
# ----------------------------------------------------------

for i in range(grid_size):

    for j in range(grid_size):

        # créer dictionnaire actions pour chaque case
        Q[(i, j)] = {}

        for a in actions:

            # valeur initiale = 0
            Q[(i, j)][a] = 0.0

# ==========================================================
# 3. PARAMÈTRES Q-LEARNING
# ==========================================================
#
# alpha  : learning rate
# gamma  : importance futur
# epsilon: exploration
#
# ==========================================================

# learning rate :
# combien on apprend à chaque étape
alpha = 0.1

# discount factor :
# importance des récompenses futures
gamma = 0.9

# exploration rate :
# probabilité de choisir action aléatoire
epsilon = 0.2

# nombre d'épisodes d'entraînement
episodes = 500

# ==========================================================
# 4. FONCTION ENVIRONNEMENT (STEP)
# ==========================================================
#
# cette fonction simule le monde
#
# entrée :
#   état + action
#
# sortie :
#   nouvel état + reward + done
#
# ==========================================================

def step(state, action):

    # ------------------------------------------------------
    # état actuel (x,y)
    # ------------------------------------------------------

    x, y = state

    # ======================================================
    # MOUVEMENT SELON ACTION
    # ======================================================

    # ------------------------------------------------------
    # aller vers le haut
    # ------------------------------------------------------

    if action == "up":

        x = max(0, x - 1)

    # ------------------------------------------------------
    # aller vers le bas
    # ------------------------------------------------------

    elif action == "down":

        x = min(grid_size - 1, x + 1)

    # ------------------------------------------------------
    # aller à gauche
    # ------------------------------------------------------

    elif action == "left":

        y = max(0, y - 1)

    # ------------------------------------------------------
    # aller à droite
    # ------------------------------------------------------

    elif action == "right":

        y = min(grid_size - 1, y + 1)

    # nouvel état après action
    new_state = (x, y)

    # ======================================================
    # SYSTÈME DE RÉCOMPENSE
    # ======================================================
    #
    # objectif atteint -> +10
    # sinon -> -1 (punition pour chaque mouvement)
    #
    # ======================================================

    if new_state == goal_state:

        reward = 10
        done = True

    else:

        reward = -1
        done = False

    return new_state, reward, done

# ==========================================================
# 5. STRATÉGIE ε-GREEDY
# ==========================================================
#
# permet équilibre :
#
# exploration vs exploitation
#
# ==========================================================

def choose_action(state):

    # ------------------------------------------------------
    # exploration (choix aléatoire)
    # ------------------------------------------------------

    if random.uniform(0, 1) < epsilon:

        return random.choice(actions)

    # ------------------------------------------------------
    # exploitation (meilleure action connue)
    # ------------------------------------------------------

    else:

        return max(

            Q[state],

            key=Q[state].get
        )

# ==========================================================
# 6. ALGORITHME Q-LEARNING
# ==========================================================
#
# boucle principale d'apprentissage
#
# ==========================================================

for episode in range(episodes):

    # état initial (start)
    state = (0, 0)

    done = False

    # ------------------------------------------------------
    # épisode = une partie complète
    # ------------------------------------------------------

    while not done:

        # choisir action
        action = choose_action(state)

        # exécuter action dans environnement
        new_state, reward, done = step(state, action)

        # ==================================================
        # FORMULE Q-LEARNING
        # ==================================================
        #
        # Q(s,a) = Q(s,a) +
        #          alpha * [reward
        #          + gamma * max(Q(s'))
        #          - Q(s,a)]
        #
        # ==================================================

        Q[state][action] += alpha * (

            reward +

            gamma * max(Q[new_state].values()) -

            Q[state][action]
        )

        # mise à jour état
        state = new_state

    # affichage progression
    if episode % 100 == 0:

        print(f"Episode {episode} terminé")

# ==========================================================
# 7. POLITIQUE OPTIMALE
# ==========================================================
#
# politique = meilleure action par état
#
# ==========================================================

print("\n===== POLITIQUE OPTIMALE =====")

policy = {}

for state in Q:

    # choisir action avec max Q-value
    best_action = max(

        Q[state],

        key=Q[state].get
    )

    policy[state] = best_action

print(policy)

# ==========================================================
# 8. TEST DE L'AGENT
# ==========================================================
#
# on suit la politique apprise
# pour aller du départ au but
#
# ==========================================================

state = (0, 0)

path = [state]

while state != goal_state:

    action = policy[state]

    state, _, _ = step(state, action)

    path.append(state)

print("\n===== CHEMIN =====")

print(path)
```
